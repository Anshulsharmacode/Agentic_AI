gemini using init and chat model

```python
from langchain.chat_models import init_chat_model
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
model= init_chat_model("google_genai:gemini-2.5-flash")
model.invoke("What is the capital of France?")
```

Stream

```python
for chunk in model.stream("write eassy in 200 word "):
    print(chunk.content, end="")
```

Batch

```python
for response in model.batch([
    "What is the capital of France?",
    "What is the capital of Germany?"
]):
    print(response.content)
```

Tool Calling

```python
from langchain.tools import tool
```

```python
@tool
def get_current_weather(location: str) -> str:
    """Get the current weather in a given location"""
    return "The current weather in {location} is sunny with a high of 25 degrees Celsius."
```

```python
model = model.bind_tools([get_current_weather])
```

```python
model.invoke("What is the current weather in New York?")
```
