Pydantic


```python
from langchain_google_genai import ChatGoogleGenerativeAI
from pydantic import BaseModel , Field
```

```python
model = ChatGoogleGenerativeAI(
    model= "gemini-2.5-flash",
    temperature=0.7
)
```

```python
class Car(BaseModel):
    name:str=Field(description="this is car name ")
    model:str=Field(description="this is model")
    price:int = Field(description="this is car price ")
    
```

```python
model_with_stru = model.with_structured_output(Car)
model_with_stru
```

```python
model_with_stru.invoke("give me name of expensive car in india")
```

nested

```python
class Actor(BaseModel):
    actor:str= Field(description="this is actor name , he is hero  "),
    villain:str = Field(description="this is villain against the actor")
    
class Movie(BaseModel):
    movie:str=Field(description="this is movie name "),
    character:list[Actor]
    year:str=Field(description="this is year of movie launch ")
```

```python
response = model.with_structured_output(Movie)
```

```python
response.invoke("what is movie conduct my sharuk khan")
```
